```python
from PIL import Image
import numpy as np
import pandas as pd
import os

def process_images(folder_path, output_path):
    # Inisialisasi writer Excel sekali untuk file output
    excel_path = os.path.join(output_path, 'image_data_analysis.xlsx')
    with pd.ExcelWriter(excel_path) as writer:
        for filename in os.listdir(folder_path):
            if filename.endswith(('.png', '.jpg', '.jpeg')):
                image_path = os.path.join(folder_path, filename)
                image = Image.open(image_path)

                # Pastikan gambar dalam format RGB
                if image.mode != 'RGB':
                    image = image.convert('RGB')

                image = image.resize((128, 128))

                # Ekstrak channel RGB dan konversi ke numpy array yang diratakan (flatten)
                r, g, b = image.split()
                r_arr = np.array(r).flatten()
                g_arr = np.array(g).flatten()
                b_arr = np.array(b).flatten()

                # Gabungkan data RGB untuk joint distribution
                rgb_arr = np.vstack([r_arr, g_arr, b_arr]).T

                # Tetapkan jumlah bins ke 9 untuk setiap channel
                bins = [9, 9, 9]

                # Hitung distribusi gabungan untuk tiga variabel RGB
                rgb_freq, edges = np.histogramdd(rgb_arr, bins=bins, density=True)

                # Ubah distribusi gabungan menjadi DataFrame untuk penyimpanan yang lebih mudah
                rgb_freq_flat = rgb_freq.flatten()
                df_rgb_joint = pd.DataFrame(rgb_freq_flat, columns=['RGB_joint_freq'])

                # Gunakan nama file sebagai nama sheet untuk menyimpan data ke Excel
                sheet_name = os.path.splitext(filename)[0]  # Menghapus ekstensi file
                df_rgb_joint.to_excel(writer, sheet_name=sheet_name[:31])  # Nama sheet dibatasi 31 karakter

                print(f'Processed and saved data from: {filename} into sheet: {sheet_name[:31]}')

# Tentukan folder tempat gambar disimpan dan folder output untuk file Excel
folder_path = '/content/drive/MyDrive/Colab Notebooks/gambar bunga byk'
output_path = '/content/drive/MyDrive/Colab Notebooks/kumpulan excel'
process_images(folder_path, output_path)
```

```python
import numpy as np
import pandas as pd
from PIL import Image
import cv2
import os
from scipy.spatial.distance import jensenshannon

def calculate_joint_distribution(image):
    if image.mode != 'RGB':
        image = image.convert('RGB')
    image = image.resize((128, 128))
    r, g, b = image.split()
    edges = cv2.Canny(np.array(image), 100, 200)

    # Menggabungkan semua channel dan tepi menjadi satu array
    features = np.stack([np.array(channel).flatten() for channel in [r, g, b, edges]], axis=1)
    # Hitung histogram dengan 9 bins untuk setiap fitur dan normalisasi
    hist = np.histogramdd(features, bins=4*[9], range=4*[(0, 255)], density=True)[0].flatten()
    return hist

def js_divergence(X, Y):
    # Menghitung JS Divergence antara dua distribusi
    return jensenshannon(X, Y, base=2)

def k_means_from_scratch(folder_path, initial_images, max_iter=100):
    # Inisialisasi centroid dari gambar yang ditentukan
    centroids = np.array([calculate_joint_distribution(Image.open(os.path.join(folder_path, img))) for img in initial_images])

    # Iterasi K-Means
    for iteration in range(max_iter):
        clusters = [[] for _ in range(len(centroids))]
        # Langkah 1: Penugasan klaster
        for filename in os.listdir(folder_path):
            if filename.endswith(('.png', '.jpg', '.jpeg')):
                image_path = os.path.join(folder_path, filename)
                image = Image.open(image_path)
                distribution = calculate_joint_distribution(image)
                distances = [js_divergence(distribution, centroid) for centroid in centroids]
                cluster_index = np.argmin(distances)
                clusters[cluster_index].append((filename, distribution))

        # Langkah 2: Pembaruan Centroid
        new_centroids = []
        for cluster in clusters:
            if cluster:  # Pastikan klaster tidak kosong
                cluster_distributions = np.array([member[1] for member in cluster])
                new_centroid = np.mean(cluster_distributions, axis=0)
                new_centroids.append(new_centroid)
            else:  # Jika klaster kosong, ambil sampel acak dari data sebagai centroid baru
                new_centroids.append(centroids[np.random.randint(len(centroids))])
        new_centroids = np.array(new_centroids)

        # Periksa konvergensi
        if np.allclose(centroids, new_centroids):
            break
        centroids = new_centroids

    # Menghasilkan mapping filename ke cluster dan jarak ke centroid
    results = {}
    for cluster_index, cluster in enumerate(clusters):
        for filename, distribution in cluster:
            distance = js_divergence(distribution, centroids[cluster_index])
            results[filename] = {'cluster': cluster_index, 'js_divergence': distance}

    return results

# Tentukan path folder dan nama-nama file untuk centroid awal
folder_path = '/content/drive/MyDrive/Colab Notebooks/gambar bunga byk'
initial_images = [f'{i} (1).png' for i in range(1, 21)]

# Jalankan K-Means dari awal
results = k_means_from_scratch(folder_path, initial_images)

# Simpan hasil ke Excel
output_path = '/content/drive/MyDrive/Colab Notebooks/kumpulan excel'
df_results = pd.DataFrame.from_dict(results, orient='index')
df_results.to_excel(os.path.join(output_path, 'kmeans_js_divergence_results_final.xlsx'))

print("K-Means clustering with JS divergence completed and results saved.")
```

```python
import numpy as np
import pandas as pd
from PIL import Image
import cv2
import os
from scipy.spatial.distance import jensenshannon

def calculate_joint_distribution(image):
    if image.mode != 'RGB':
        image = image.convert('RGB')
    image = image.resize((128, 128))
    r, g, b = image.split()
    edges = cv2.Canny(np.array(image), 100, 200)

    features = np.stack([np.array(channel).flatten() for channel in [r, g, b, edges]], axis=1)
    hist = np.histogramdd(features, bins=4*[9], range=4*[(0, 255)], density=True)[0].flatten()
    return hist

def k_means_from_scratch(folder_path, initial_images, max_iter=100):
    centroids = np.array([calculate_joint_distribution(Image.open(os.path.join(folder_path, img))) for img in initial_images])
    clusters = [[] for _ in range(len(centroids))]

    for iteration in range(max_iter):
        clusters = [[] for _ in range(len(centroids))]
        for filename in os.listdir(folder_path):
            if filename.endswith(('.png', '.jpg', '.jpeg')):
                image_path = os.path.join(folder_path, filename)
                image = Image.open(image_path)
                distribution = calculate_joint_distribution(image)
                distances = [jensenshannon(distribution, centroid, base=2) for centroid in centroids]
                cluster_index = np.argmin(distances)
                clusters[cluster_index].append((filename, distribution))

        new_centroids = []
        for cluster in clusters:
            if cluster:
                cluster_distributions = np.array([member[1] for member in cluster])
                new_centroid = np.mean(cluster_distributions, axis=0)
                new_centroids.append(new_centroid)
            else:
                new_centroids.append(centroids[np.random.randint(len(centroids))])
        new_centroids = np.array(new_centroids)

        if np.allclose(centroids, new_centroids):
            break
        centroids = new_centroids

    # Menghitung JS divergence final antara centroid dan gambar
    final_js_divergences = {}
    for i, centroid in enumerate(centroids):
        js_divergences = {}
        for cluster in clusters[i]:
            filename, distribution = cluster
            js_distance = jensenshannon(distribution, centroid, base=2)
            js_divergences[filename] = js_distance
        final_js_divergences[f'Centroid_{i+1}'] = js_divergences

    return centroids, clusters, final_js_divergences

folder_path = '/content/drive/MyDrive/Colab Notebooks/gambar bunga byk'
initial_images = [f'{i} (1).png' for i in range(1, 21)]

centroids, clusters, final_js_divergences = k_means_from_scratch(folder_path, initial_images)

# Menyimpan JS divergence final ke Excel
output_path = '/content/drive/MyDrive/Colab Notebooks/kumpulan excel'
with pd.ExcelWriter(os.path.join(output_path, 'kmeans_js_divergence_results_finalR1.xlsx')) as writer:
    for centroid_name, js_divergences in final_js_divergences.items():
        df_js = pd.DataFrame.from_dict(js_divergences, orient='index', columns=['JS_Divergence'])
        df_js.to_excel(writer, sheet_name=centroid_name)

print("K-Means clustering with JS divergence completed and final results saved.")
```
